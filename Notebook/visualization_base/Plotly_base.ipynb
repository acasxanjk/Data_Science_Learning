```python
from die import Die
import plotly.express as px
```

```python
#Create two D6
die_1 = Die()
die_2 = Die()

#Roll the dices for several times and store the results in a list
results = []
for roll_num in range(1000):
    result = die_1.roll() + die_2.roll()
    results.append(result)

print(results)
```

```python
#Analyst the results
frequencies = []
max_results = die_1.num_sides + die_2.num_sides
poss_results = range(2,max_results+1)
for value in poss_results:
    frequency = results.count(value)
    frequencies.append(frequency)

print(frequencies)
```

```python
#Results visualization
title = "Results of Rolling Two D6 1,000 Times"
labels = {"x":"Results","y":"Frequency of Results"}
fig = px.bar(x=poss_results, y=frequencies,title=title,labels=labels)

#Further customization of graphics
fig.update_layout(xaxis_dtick=1)
fig.show()
```

```python
#Save the picture in HTML format
fig.write_html('Plotly_base.html')
```
