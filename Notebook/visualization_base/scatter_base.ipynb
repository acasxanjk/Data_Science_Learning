```python
import matplotlib.pyplot as plt
#choose the styles in the matplot
plt.style.available
```

```python
plt.style.use('seaborn-v0_8-bright')
fig,ax = plt.subplots()
ax.scatter(2,4,s=200)

#set the title and add labels to the axes
ax.set_title('Square Numbers',fontsize=24)
ax.set_xlabel('Value',fontsize=12)
ax.set_ylabel('Square of Value',fontsize=12)

#set the style of the labels
ax.tick_params(labelsize=12)

plt.show()
```

```python
x_values = [1,2,3,4,5]
y_values = [1,4,9,16,25]

plt.style.use('seaborn-v0_8-bright')
fig, ax = plt.subplots()
ax.scatter(x_values,y_values,s=100)
```

```python
x_values = range(1,1001)
y_values = [x**2 for x in x_values]

plt.style.use('seaborn-v0_8')
fig,ax = plt.subplots()
ax.scatter(x_values,y_values,s=10,c=y_values,cmap=plt.cm.Reds)#use the colormap to show the increse

#set title and set labels to the axes
ax.set_title('Square Numbers',fontsize=24)
ax.set_xlabel('Value',fontsize=12)
ax.set_ylabel('Square of Value',fontsize=12)

#set axis range
ax.axis([0,1100,0,1_100_000])

#Custom scale markings
ax.ticklabel_format(style='plain')

plt.show()
```
